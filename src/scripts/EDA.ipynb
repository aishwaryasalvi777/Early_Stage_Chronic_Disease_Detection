# 1. Row counts and memory footprint

```python
# 1. Row counts and memory footprint
print("=" * 60)
print("ROW COUNTS & MEMORY FOOTPRINT")
print("=" * 60)

dfs = {'patients': patients, 'diagnoses': diagnoses, 'lab_results': lab_results, 
       'medications': medications, 'outcomes': outcomes}

for name, df in dfs.items():
    memory_mb = df.memory_usage(deep=True).sum() / 1024**2
    print(f"{name:15} | Rows: {len(df):>10,} | Memory: {memory_mb:>8.2f} MB")
```

# 2. Missing value patterns by table and column

```python
# 2. Missing value patterns by table and column (table output)
summary_rows = []
column_rows = []

for name, df in dfs.items():
    total_rows = len(df)
    total_cols = len(df.columns)
    total_cells = total_rows * total_cols
    missing_cells = int(df.isna().sum().sum())
    missing_pct = (missing_cells / total_cells * 100) if total_cells else 0.0

    summary_rows.append({
        "table": name,
        "rows": total_rows,
        "columns": total_cols,
        "total_cells": total_cells,
        "missing_cells": missing_cells,
        "missing_pct": round(missing_pct, 4)
    })

    missing_by_col = df.isna().sum()
    missing_by_col = missing_by_col[missing_by_col > 0].sort_values(ascending=False)

    for col_name, miss_count in missing_by_col.items():
        column_rows.append({
            "table": name,
            "column": col_name,
            "missing_count": int(miss_count),
            "missing_pct_in_table": round((miss_count / total_rows * 100), 4)
        })

missing_summary_df = pd.DataFrame(summary_rows).sort_values("missing_cells", ascending=False)
missing_columns_df = pd.DataFrame(column_rows).sort_values(["missing_count", "table"], ascending=[False, True])

print("Missing summary by table")
display(missing_summary_df)

print("Missing values by column (only columns with missing values)")
if missing_columns_df.empty:
    print("No missing values in any column across all tables.")
else:
    display(missing_columns_df)
```

```python
# 3. Duplicate patient IDs or records
print("\n" + "=" * 60)
print("DUPLICATE ANALYSIS")
print("=" * 60)

for name, df in dfs.items():
    if 'patient_id' in df.columns:
        unique_patients = df['patient_id'].nunique()
        total_rows = len(df)
        
        # Check for exact duplicates
        exact_dupes = df.duplicated().sum()
        
        print(f"\n{name.upper()}:")
        print(f"  Total rows: {total_rows:,}")
        print(f"  Unique patient_ids: {unique_patients:,}")
        print(f"  Exact row duplicates: {exact_dupes:,}")
        
        if name == 'patients':  # patients should be 1 row per patient
            if unique_patients == total_rows:
                print(f"  ✓ OK: 1 row per patient")
            else:
                print(f"  ⚠ WARNING: Multiple rows per patient")
```

```python
# 4. Date ranges for each table (temporal coverage)
print("\n" + "=" * 60)
print("TEMPORAL COVERAGE (DATE RANGES)")
print("=" * 60)

date_columns = {
    'diagnoses': 'visit_date',
    'lab_results': 'test_date',
    'medications': 'start_date',
    'outcomes': 'admission_date'
}

for name, date_col in date_columns.items():
    df = dfs[name]
    if date_col in df.columns:
        df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
        min_date = df[date_col].min()
        max_date = df[date_col].max()
        
        print(f"\n{name.upper()} ({date_col}):")
        print(f"  From: {min_date.date()}")
        print(f"  To:   {max_date.date()}")
        print(f"  Span: {(max_date - min_date).days} days ({(max_date - min_date).days / 365.25:.2f} years)")
```

```python
# 5. Identify patients with no diagnosis/outcome records (data completeness by patient)
print("\n" + "=" * 60)
print("DATA COMPLETENESS BY PATIENT")
print("=" * 60)

total_patients = len(patients)
print(f"\nTotal patients: {total_patients:,}")

# Patients with diagnoses
patients_with_diagnoses = diagnoses['patient_id'].nunique()
patients_without_diagnoses = total_patients - patients_with_diagnoses
print(f"\nPatients WITH diagnoses: {patients_with_diagnoses:,} ({100*patients_with_diagnoses/total_patients:.2f}%)")
print(f"Patients WITHOUT diagnoses: {patients_without_diagnoses:,} ({100*patients_without_diagnoses/total_patients:.2f}%)")

# Patients with outcomes
patients_with_outcomes = outcomes['patient_id'].nunique()
patients_without_outcomes = total_patients - patients_with_outcomes
print(f"\nPatients WITH outcomes: {patients_with_outcomes:,} ({100*patients_with_outcomes/total_patients:.2f}%)")
print(f"Patients WITHOUT outcomes: {patients_without_outcomes:,} ({100*patients_without_outcomes/total_patients:.2f}%)")

# Patients with labs
patients_with_labs = lab_results['patient_id'].nunique()
patients_without_labs = total_patients - patients_with_labs
print(f"\nPatients WITH lab_results: {patients_with_labs:,} ({100*patients_with_labs/total_patients:.2f}%)")
print(f"Patients WITHOUT lab_results: {patients_without_labs:,} ({100*patients_without_labs/total_patients:.2f}%)")

# Patients with medications
patients_with_meds = medications['patient_id'].nunique()
patients_without_meds = total_patients - patients_with_meds
print(f"\nPatients WITH medications: {patients_with_meds:,} ({100*patients_with_meds/total_patients:.2f}%)")
print(f"Patients WITHOUT medications: {patients_without_meds:,} ({100*patients_without_meds/total_patients:.2f}%)")
```
